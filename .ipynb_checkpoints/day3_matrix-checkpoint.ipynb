```python
import numpy as np
import matplotlib.pyplot as plt

# 如果需要画图
%matplotlib inline
```

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])

AB = A @ B
print(AB)
```

```python
AB_manual = np.zeros((A.shape[0], B.shape[1]))
for i in range(A.shape[0]):
    for j in range(B.shape[1]):
        for k in range(A.shape[1]):
            AB_manual[i][j] += A[i][k] * B[k][j]

print(AB_manual)
print("一致:", np.allclose(AB, AB_manual))
```

```python
BA = B @ A
print("A @ B 形状:", AB.shape)
print("B @ A 形状:", BA.shape)
print(BA)
```

```python
M = np.array([[4, 1], [2, 3]])
eigvals, eigvecs = np.linalg.eig(M)
print("特征值:", eigvals)
print("特征向量:", eigvecs)
```

```python
for i in range(len(eigvals)):
    v = eigvecs[:, i]
    lam = eigvals[i]
    lhs = M @ v
    rhs = lam * v
    print(f"λ{i} = {lam}")
    print(f"  M@v = {lhs}")
    print(f"  λv  = {rhs}")
    print(f"  相等: {np.allclose(lhs, rhs)}")
```

```python
np.random.seed(42)
A3 = np.random.randn(3, 3)
b3 = np.random.randn(3)

print("A3 =")
print(A3)
print("b3 =", b3)

x = np.linalg.solve(A3, b3)
print("\n解 x =", x)
print("验证 A3 @ x =", A3 @ x)
print("是否等于 b3:", np.allclose(A3 @ x, b3))
```

```python
x_inv = np.linalg.inv(A3) @ b3
print("用求逆解 x =", x_inv)
print("和 solve 结果是否一致:", np.allclose(x, x_inv))
```
